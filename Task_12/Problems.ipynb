**Problem 5**

```python
n2 = int(input()) # user input but avoiding redunduncy in variables
np5 = list(map(int, input().split()))
p1_sum, p2_sum = 0,0
left, right = 0, n2-1
p1_turn = True

while left <= right:
    if np5[left] > np5[right]:  # Choose the larger card between the left and right
        card = np5[left]
        left += 1
    else:
        card = np5[right]
        right -= 1

    if p1_turn:  
        p1_sum += card
    else:  
        p2_sum += card
    
    p1_turn = not p1_turn  # Switch turns

# Output the final scores
print(f"{p1_sum} {p2_sum}")
```
